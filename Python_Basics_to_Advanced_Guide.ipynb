## Q21. What is NumPy?

```python
import numpy as np
```

## Q23. How to create a NumPy array?

## array() converts a Python list into a NumPy array, which supports fast mathematical operations.

```python
import numpy as np
arr = np.array([10, 20, 30])
print(arr)
```

## Q24. What is the difference between a Python list and a NumPy array?

## A NumPy array is faster and supports element-wise mathematical operations directly, while a normal Python list does not.

```python
arr = np.array([1, 2, 3])
print(arr * 2) # works on every element
```

## Q25. How to create a 2D array (matrix) in NumPy?

## Pass a list of lists to array() to create a 2D array, like a table with rows and columns.

```python
matrix = np.array([[1, 2], [3, 4]])
print(matrix)
```

## Q26. How to check the shape of a NumPy array?

## shape shows the number of rows and columns in the array, as (rows, columns).

```python
matrix = np.array([[1, 2], [3, 4]])
print(matrix.shape)
```

## Q27. How to create an array of zeros?

## zeros() creates an array filled with 0s of the given shape - useful to initialize data.

```python
np.zeros((2, 3))
```

## Q28. How to create an array of ones?

## ones() creates an array filled with 1s of the given shape.

```python
np.ones((2, 2))
```

## Q29. How to create a sequence of numbers in NumPy?

## arange() generates numbers in a range, similar to Python's range(), but returns a NumPy array.

```python
np.arange(0, 10, 2)
```

## Q30. How to create evenly spaced numbers between two values?

## linspace() generates a fixed number of evenly spaced values between a start and end point.

```python
np.linspace(0, 1, 5)
```

## Q31. How to find the mean of a NumPy array?

## mean() calculates the average of all values in the array.

```python
arr = np.array([10, 20, 30])
print(arr.mean())
```

## Q32. How to find the sum of a NumPy array?

## sum() adds up all the values in the array.

```python
arr = np.array([10, 20, 30])
print(arr.sum())
```

## Q33. How to find the max and min value in an array?

## max() and min() return the largest and smallest values in the array.

```python
arr = np.array([10, 20, 30])
print(arr.max(), arr.min())
```

## Q34. How to find the standard deviation in NumPy?

## std() calculates how spread out the values are from the mean.

```python
arr = np.array([10, 20, 30])
print(arr.std())
```

## Q35. How to reshape a NumPy array?

## reshape() changes the shape (rows and columns) of an array without changing its data.

```python
arr = np.arange(6)
print(arr.reshape(2, 3))
```

## Q36. How to select elements from a NumPy array using conditions?

## Just like Pandas filtering, you can pass a condition inside square brackets to get matching elements.

```python
arr = np.array([10, 20, 30, 40])
print(arr[arr > 20])
```

## Q37. How to generate random numbers in NumPy?

## np.random.rand() generates random decimal numbers between 0 and 1.

```python
np.random.rand(3)
```

## Q38. How to generate random integers in NumPy?

## np.random.randint() generates random whole numbers within a given range.

```python
np.random.randint(1, 100, 5)
```

## Q39. What is broadcasting in NumPy?

## Broadcasting lets NumPy apply an operation between arrays of different shapes automatically, without writing a loop.

```python
arr = np.array([1, 2, 3])
print(arr + 10)
```

## Q40. How to find the transpose of a matrix in NumPy?

## T flips a matrix - rows become columns and columns become rows.

```python
matrix = np.array([[1, 2], [3, 4]])
print(matrix.T)
```

## Q41. How to stack two NumPy arrays together?

## np.concatenate() or np.vstack()/np.hstack() join arrays vertically or horizontally.

```python
a = np.array([1, 2])
b = np.array([3, 4])
print(np.concatenate([a, b]))
```

## Q42. How to check the data type of a NumPy array?

## dtype shows what kind of values the array holds, like int64 or float64.

```python
arr = np.array([1, 2, 3])
print(arr.dtype)
```

## Q43. How to find unique values in a NumPy array?

## np.unique() returns all distinct values from the array, sorted and without duplicates

```python
arr = np.array([1, 2, 2, 3, 3, 3])
print(np.unique(arr))
```

## Q44. How to sort a NumPy array?

## np.sort() arranges the array values in ascending order.

```python
arr = np.array([30, 10, 20])
print(np.sort(arr))
```

## Q45. Why is NumPy faster than normal Python lists?

## NumPy arrays store data in a fixed, compact format in memory and use optimized C code internally, making calculations much faster than looping through a normal list.
